```python
import pandas as pd
df=pd.read_csv('customer_shopping_behaviour.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df.columns
```

```python
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
#feature engineering:create a new column
labels=['Teenage','Adult','Middle-Aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
df[['age','age_group']].head(30)
```

```python
#mapping
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Annually':365,
    'Bi-Weekly':14,
    'Every 3 Months':90
}

df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
df[['purchase_frequency_days','frequency_of_purchases']].head(30)
```

```python
(df['discount_applied']==df['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used', axis=1)
df.columns
```

```python
df.to_csv("customer_shopping_behaviour_preprocessed.csv", index=False)
print("Preprocessed dataset saved successfully!")
```
